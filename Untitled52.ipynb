````python
import hashlib as hasher

class Block:
  def __init__(self, index, timestamp, data, previous_hash):
    self.index = index
    self.timestamp = timestamp
    self.data = data
    self.previous_hash = previous_hash
    self.hash = self.hash_block()

  def hash_block(self):
    sha = hasher.sha256()
    combined_string = (str(self.index) +
                       str(self.timestamp) +
                       str(self.data) +
                       str(self.previous_hash))
    sha.update(combined_string.encode('utf-8'))
    return sha.hexdigest()
````

````python
import datetime as date

def create_genesis_block():
  # Manually construct a block with
  # index zero and arbitrary previous hash
  return Block(0, date.datetime.now(), "Genesis Block", "0")
````

````python
def next_block(last_block):
  this_index = last_block.index + 1
  this_timestamp = date.datetime.now()
  this_data = "Hey! I'm block " + str(this_index)
  this_hash = last_block.hash
  return Block(this_index, this_timestamp, this_data, this_hash)
````

````python
# Create the blockchain and add the genesis block
blockchain = [create_genesis_block()]
previous_block = blockchain[0]

# How many blocks should we add to the chain
# after the genesis block
num_of_blocks_to_add = 20

# Add blocks to the chain
for i in range(0, num_of_blocks_to_add):
  block_to_add = next_block(previous_block)
  blockchain.append(block_to_add)
  previous_block = block_to_add
  # Tell everyone about it!
  print("Block #{} has been added to the blockchain!".format(block_to_add.index))
  print("Hash: {}\n".format(block_to_add.hash))
````

### Proof-of-Work Blockchain Implementation

Below, we subclass our original `Block` (or create an updated version) to include `difficulty` and a `nonce` for mining.

````python
import time
import hashlib as hasher

class PoWBlock:
  def __init__(self, index, timestamp, data, previous_hash, difficulty):
    self.index = index
    self.timestamp = timestamp
    self.data = data
    self.previous_hash = previous_hash
    self.difficulty = difficulty
    self.nonce = 0
    self.hash = self.mine_block()

  def calculate_hash(self):
    sha = hasher.sha256()
    combined_string = (str(self.index) +
                       str(self.timestamp) +
                       str(self.data) +
                       str(self.previous_hash) +
                       str(self.nonce))
    sha.update(combined_string.encode('utf-8'))
    return sha.hexdigest()

  def mine_block(self):
    # The hash must start with a specific number of zeros dictated by difficulty
    target = '0' * self.difficulty
    while True:
      current_hash = self.calculate_hash()
      if current_hash.startswith(target):
        print(f"Block #{self.index} successfully mined with nonce {self.nonce}!")
        print(f"Hash: {current_hash}\n")
        return current_hash
      self.nonce += 1
````

Let's test our mining mechanism with a low difficulty level (e.g., 4 leading zeros) so it computes quickly in our notebook environment.

````python
# Initialize the PoW blockchain with a difficulty of 4 leading zeros
difficulty = 4
pow_blockchain = []

# Create Genesis Block
genesis_pow = PoWBlock(0, time.time(), "Genesis Block with PoW", "0", difficulty)
pow_blockchain.append(genesis_pow)

# Mine a few subsequent blocks
previous_block = pow_blockchain[0]
num_pow_blocks = 3

for i in range(1, num_pow_blocks + 1):
  start_time = time.time()
  new_block = PoWBlock(
      index=i,
      timestamp=time.time(),
      data=f"Transaction Data for Block {i}",
      previous_hash=previous_block.hash,
      difficulty=difficulty
  )
  end_time = time.time()
  pow_blockchain.append(new_block)
  previous_block = new_block
  print(f"Time taken to mine Block #{i}: {end_time - start_time:.4f} seconds\n")
````

### Visualizing the Relationship Between Difficulty and Mining Time

### Validating Blockchain Integrity

To ensure our PoW blockchain remains secure and untampered with, we will implement a validation function. This checks both the cryptographic linkages between blocks and verifies that the PoW target was correctly met for each block.

````python
def is_chain_valid(chain):
    for i in range(1, len(chain)):
        current_block = chain[i]
        previous_block = chain[i-1]

        # 1. Recalculate hash and check if it matches stored hash
        if current_block.hash != current_block.calculate_hash():
            print(f"[INVALID] Block #{current_block.index} hash has been tampered with!")
            return False

        # 2. Check if the chain linkage is intact
        if current_block.previous_hash != previous_block.hash:
            print(f"[INVALID] Block #{current_block.index} previous hash mismatch!")
            return False

        # 3. Verify that the Proof-of-Work target difficulty was satisfied
        target = '0' * current_block.difficulty
        if not current_block.hash.startswith(target):
            print(f"[INVALID] Block #{current_block.index} does not meet difficulty requirements!")
            return False

    print("[SUCCESS] The blockchain integrity is valid!")
    return True

# Validate the existing pow_blockchain
is_chain_valid(pow_blockchain)
````

### Dynamic Difficulty Adjustment Algorithm

In real blockchain networks (like Bitcoin), the hashing power of the network fluctuates. To maintain a stable block generation rate, the network periodically adjusts the mining `difficulty`:
- If blocks are mined **faster** than the target block time, **increase** the difficulty.
- If blocks are mined **slower** than the target block time, **decrease** the difficulty.

````python
import time

# Target time to mine a single block (in seconds)
TARGET_BLOCK_TIME = 0.5
# How often to adjust difficulty (every N blocks)
ADJUSTMENT_INTERVAL = 3

def adjust_difficulty(last_blocks, current_difficulty):
    """
    Adjusts the difficulty based on the average mining time of the last N blocks.
    """
    if len(last_blocks) < ADJUSTMENT_INTERVAL:
        return current_difficulty

    # Calculate total time taken to mine the interval blocks
    first_block = last_blocks[-ADJUSTMENT_INTERVAL]
    last_block = last_blocks[-1]
    actual_time = last_block.timestamp - first_block.timestamp
    target_time = TARGET_BLOCK_TIME * ADJUSTMENT_INTERVAL

    print(f"\n--- Difficulty Retarget --- (Every {ADJUSTMENT_INTERVAL} blocks)")
    print(f"Actual time for last {ADJUSTMENT_INTERVAL} blocks: {actual_time:.4f}s")
    print(f"Target time: {target_time:.4f}s")

    # Prevent division by zero and extreme fluctuations
    if actual_time < target_time / 2:
        new_difficulty = current_difficulty + 1
        print(f"Mining too fast! Increasing difficulty: {current_difficulty} -> {new_difficulty}")
    elif actual_time > target_time * 2:
        new_difficulty = max(1, current_difficulty - 1)
        print(f"Mining too slow! Decreasing difficulty: {current_difficulty} -> {new_difficulty}")
    else:
        new_difficulty = current_difficulty
        print(f"Mining pace is optimal. Keeping difficulty at: {new_difficulty}")
    print("---------------------------\n")

    return new_difficulty
````

````python
# Simulation of mining with dynamic difficulty adjustment
current_difficulty = 3
dynamic_blockchain = []

# Generate genesis block
genesis = PoWBlock(0, time.time(), "Genesis", "0", current_difficulty)
dynamic_blockchain.append(genesis)

# Mine 9 blocks and see the difficulty adapt
for i in range(1, 10):
    start_time = time.time()

    # Check and adjust difficulty every ADJUSTMENT_INTERVAL blocks
    if i % ADJUSTMENT_INTERVAL == 0:
        current_difficulty = adjust_difficulty(dynamic_blockchain, current_difficulty)

    new_block = PoWBlock(
        index=i,
        timestamp=time.time(),
        data=f"Block {i} Data",
        previous_hash=dynamic_blockchain[-1].hash,
        difficulty=current_difficulty
    )

    dynamic_blockchain.append(new_block)
    elapsed = time.time() - start_time
    print(f"Block #{i} mined in {elapsed:.4f}s using difficulty {current_difficulty}")
````

A hash function takes an input of any size (letters, numbers, or entire files) and outputs a fixed-size string of characters, which looks completely random.

Since anyone can write data to a database, how do we prevent malicious users from spamming the network with fake blocks? We introduce a computational challenge: **Proof-of-Work**.

### The Mining Puzzle:
Instead of letting blocks be added instantly, the network demands that miners solve a puzzle. The puzzle is to find a number called a **Nonce** (number used once) which, when combined with the block data, produces a hash starting with a specific number of leading zeros (the **Difficulty**).

$$\text{SHA-256}(\text{Index} + \text{Timestamp} + \text{Data} + \text{Previous Hash} + \mathbf{Nonce}) < \text{Target}$$

---

## 📈 Chapter 3: Understanding Mining Curves

Why does adding just one zero to the difficulty make mining take so much longer?

Because hashes are in hexadecimal ($0-9$ and $a-f$), there are $16$ possible values for each character. To get a hash that starts with:
- **1 zero**: $1$ in $16$ chance ($16^1$ average guesses needed)
- **2 zeros**: $1$ in $256$ chance ($16^2$ average guesses needed)
- **5 zeros**: $1$ in $1,048,576$ chance ($16^5$ average guesses needed)

This creates an **exponential curve**, which we verified in our earlier benchmark plotting exercise. The effort scales exponentially, while validation remains instantaneous (taking just $1$ calculation)!

---

## ⚙️ Chapter 4: Dynamic Difficulty Adjustments (Retargeting)

If more miners join the network with faster computers, blocks will be mined too quickly. If miners leave, blocks will take too long.

To solve this, we implement a **Difficulty Adjustment Algorithm** (like Bitcoin's retargeting mechanism):

$$\text{New Difficulty} = \text{Current Difficulty} \times \left( \frac{\text{Actual Time Taken}}{\text{Target Time}} \right)$$

If the actual elapsed mining time is less than our target window, the difficulty increments, scaling up the computational workload to keep block production times perfectly consistent.

## 🔑 Chapter 1: The Cryptographic Foundations of SHA-256

How do we guarantee that data in a decentralized ledger is authentic, unalterable, and easily verifiable by anyone in the world without a central authority? We rely on **cryptography**—specifically, a mathematical tool called a cryptographic hash function.

---

### 📌 Section 1: What is a Cryptographic Hash?

Imagine a machine that takes any object—a letter, a book, or an entire digital database—shreds it down, analyzes its atomic structure, and outputs a unique, fixed-size physical stamp. No matter how massive or tiny the original input is, the resulting stamp is always exactly the same length.

In computer science, **SHA-256 (Secure Hash Algorithm 256-bit)** is that machine. It accepts an input of any size and produces a **256-bit (32-byte) signature**, typically expressed as a 64-character hexadecimal string.

---

### 📌 Section 2: The Core Mathematical Guarantees

To be useful for securing billions of dollars in digital assets, a hash function must satisfy several rigorous mathematical constraints:

#### 1. Deterministic
A hash function is perfectly predictable. If you pass the word `"Bitcoin"` into SHA-256 today, tomorrow, or in a hundred years, it will **always** yield the exact same hex string. Without this consistency, nodes in a blockchain network could never agree on the state of the ledger.

#### 2. Pre-image Resistance (One-Way Function)
You can easily turn an input into a hash, but you cannot reconstruct the input using the hash. It is a computational dead-end.
* **Input $\rightarrow$ Hash**: Instantaneous (microseconds).
* **Hash $\rightarrow$ Input**: Impossible (would require brute-forcing $2^{256}$ combinations, which would take supercomputers longer than the age of the universe).

#### 3. Collision Resistance
It is practically impossible to find two different inputs that produce the exact same output hash ($A \neq B$, but $\text{Hash}(A) = \text{Hash}(B)$). If a hash function is not collision-resistant, bad actors could swap out valid transaction data with malicious data without altering the overall block signature.

#### 4. High Avalanche Effect
This is the most visible security mechanism of hashing. A minute change in the input (e.g., swapping a capital letter to lowercase, or adding a single period) causes an entirely catastrophic, unpredictable change in the output. The resulting hash looks completely randomized and shares no visual relationship with the original version.

## ⛏️ Chapter 2: Proof-of-Work and the Mining Puzzle

In a decentralized network, a single authority decides what transactions are valid. But in a decentralized network where anyone can participate, how do we prevent malicious actors from spamming the network, rewriting history, or validating fake transactions?

We introduce a computational gatekeeper: **Consensus Mechanisms**—specifically, **Proof-of-Work (PoW)**.

---

### 📌 Section 1: The Core Mechanism

Proof-of-Work requires network participants (called **miners**) to expend measurable computational effort before their proposed block is accepted by the network.

Instead of checking if someone is trustworthy, the network checks if they have successfully solved a difficult mathematical puzzle. This makes writing fraud to the blockchain too expensive and economically irrational to attempt.

---

### 📌 Section 2: What is the Mining Puzzle?

To understand mining, we must look at a special variable inside a block: the **Nonce** (which stands for "number used once").

The mining puzzle is simple to state but incredibly difficult to execute:
* Find a numeric value for the `nonce` such that, when combined with the block's details (index, timestamp, data, and previous hash), the resulting SHA-256 hash starts with a specific number of leading zeros (the **Difficulty**).

$$\text{SHA-256}(\text{Index} + \text{Timestamp} + \text{Data} + \text{Previous Hash} + \mathbf{Nonce}) < \text{Target Space}$$

If the required difficulty is $4$, the first four characters of the hex string must be `0000`. Because SHA-256 is completely unpredictable, the *only* way to find this number is through brute force: trying different values for the nonce ($0, 1, 2, 3, \dots$) until a winning hash is found.

---

### 📌 Section 3: Asymmetrical Workload (Hard to Find, Easy to Verify)

One of the most elegant aspects of PoW is its asymmetry:
* **Mining (Hard)**: A miner might have to iterate through hundreds of thousands or millions of nonces to find a hash that fits the target requirement.
* **Verification (Instantaneous)**: Once a miner finds the correct nonce, they broadcast it to the network. Any other node can check the calculation in a single step by hashing the block data with that specific nonce once. If it starts with the correct number of zeros, they accept the block.

## 📈 Chapter 3: Understanding Mining Curves and Difficulty

Why does adding just a single digit of difficulty turn a fraction-of-a-second calculation into a minutes-long, electricity-guzzling computational struggle?

To understand this, we must look at how cryptographic space is structured, how probabilities scale, and why the difficulty curve is strictly **exponential**.

---

### 📌 Section 1: The Hexadecimal Probability Space

SHA-256 outputs hashes in hexadecimal (base-16) format. Each character in a hex hash can be one of 16 possible values: `0` through `9` or `a` through `f`.

When we dictate that a valid hash must start with leading zeros, we are dramatically shrinking the pool of valid outputs. Let's look at how the probability of finding a valid hash changes with every zero we demand:

* **1 Leading Zero** (`0...`):
  * Probabilistic chance: $1$ in $16^1$ ($1$ in $16$ attempts)
* **2 Leading Zeros** (`00...`):
  * Probabilistic chance: $1$ in $16^2$ ($1$ in $256$ attempts)
* **3 Leading Zeros** (`000...`):
  * Probabilistic chance: $1$ in $16^3$ ($1$ in $4,096$ attempts)
* **4 Leading Zeros** (`0000...`):
  * Probabilistic chance: $1$ in $16^4$ ($1$ in $65,536$ attempts)
* **5 Leading Zeros** (`00000...`):
  * Probabilistic chance: $1$ in $16^5$ ($1$ in $1,048,576$ attempts)

### 📌 Section 2: Visualizing the Exponential Scaling

We verified this behavior in our benchmark plotting exercise. The effort scales exponentially, while validation remains instantaneous (taking just $1$ calculation)!

---

### 📌 Section 3: The Immutability of Block History

This exponential barrier is precisely what keeps historical block data immutable. If an attacker wants to tamper with a transaction in **Block 5**, they must:
1. Alter the block data.
2. Remine Block 5 (re-solving the difficult computational puzzle with its updated hash).
3. Remine **all subsequent blocks** (Blocks 6, 7, 8, etc.), since their chain links depend directly on Block 5's hash.

To successfully rewrite history, the attacker must possess more computing power than the entire rest of the network combined—a threshold known as a **51% Attack**.

## ⚙️ Chapter 4: Dynamic Difficulty Adjustments (Retargeting)

What happens when a new supercomputer joins our blockchain network and starts mining blocks in milliseconds? Or what if half of our miners turn off their machines, causing block times to stretch to hours?

To prevent chaos, a secure blockchain must stay predictable. It does this through **Dynamic Difficulty Adjustments**.

---

### 📌 Section 1: The Principle of Stable Block Times

If block creation is too fast, the network risks splitting into alternative chain paths because nodes cannot sync block histories quickly enough. If it is too slow, users must wait hours for transaction confirmations.

To maintain an equilibrium, the protocol dynamically retargets its difficulty requirements at regular intervals based on actual hashing power.

---

### 📌 Section 2: The Retargeting Equation

Every $N$ blocks (known as the *Adjustment Interval*), the protocol evaluates the real time elapsed since the previous target epoch. It then recalibrates the workload using this proportion:

$$\text{New Difficulty Ratio} = \frac{\text{Actual Time Elapsed}}{\text{Target Time Window}}$$

If the actual elapsed time is shorter than the target, the mining difficulty increments. If it took longer, the difficulty decreases. This elegant feedback loop acts as the regulatory thermostat of the blockchain ecosystem.

---

### 📌 Section 3: Simulating the Shift in Python

Recall the dynamic mining simulation we ran earlier in this notebook:
1. We started with a base difficulty of `3`.
2. When blocks were generated faster than our `0.5` second threshold, the difficulty auto-incremented to `4` and then `5`.
3. As a result, mining time expanded back towards our target, establishing a balanced system without human intervention.

This self-regulating equilibrium is the final key element required to deploy a production-ready, globally resilient decentralized network.

### 🛠️ Hands-On: Simulate the Mining Puzzle

Run the interactive Python script below to simulate how a miner solves a Proof-of-Work puzzle. You can tweak the `difficulty_level` and the `block_data` to see how it impacts both the required number of guesses (nonces tried) and the final target hash.

````python
import hashlib
import time

def simulate_mining_puzzle(block_data, difficulty_level):
    target_prefix = '0' * difficulty_level
    nonce = 0
    start_time = time.time()

    print(f"[STARTING] Mining block with difficulty: {difficulty_level} (Target prefix: '{target_prefix}')")
    print(f"[DATA] {block_data}\n")

    while True:
        # Combine block data with the changing nonce string
        input_payload = f"{block_data}-{nonce}"
        # Generate SHA-256 Hash
        computed_hash = hashlib.sha256(input_payload.encode('utf-8')).hexdigest()

        # Check if hash matches target difficulty
        if computed_hash.startswith(target_prefix):
            duration = time.time() - start_time
            print(f"[SUCCESS] Block successfully mined!")
            print(f"-> Final Nonce tried: {nonce}")
            print(f"-> Matching Hash: {computed_hash}")
            print(f"-> Duration: {duration:.5f} seconds\n")
            return nonce, computed_hash

        nonce += 1

# --- Test the Mining Puzzle Here! ---
test_data = "Transaction: Alice pays Bob 5.5 BTC"
test_difficulty = 4  # Change this from 1 to 5 to observe exponential difficulty scaling!

simulate_mining_puzzle(test_data, test_difficulty)
````

## 📚 Appendix: Glossary of Key Terms

Here is a quick reference guide to the core cryptographic and consensus terms used throughout this interactive e-book:

* **SHA-256 (Secure Hash Algorithm 256-bit)**: A cryptographic hash function that takes an input of any size and produces a fixed-size 256-bit signature. It is deterministic, one-way, and collision-resistant.
* **Nonce (Number Used Once)**: An arbitrary number that miners change iteratively to find a block hash that meets the network's difficulty criteria.
* **Difficulty**: A dynamic value that determines how hard it is to find a hash with the required number of leading zeros. It scales exponentially with each additional zero.
* **Proof-of-Work (PoW)**: A consensus mechanism requiring network participants (miners) to solve computationally intensive mathematical puzzles to validate transactions and secure the network.
* **Dynamic Retargeting**: The automatic adjustment of the mining difficulty at regular intervals to maintain a stable, target block production rate despite fluctuations in network computing power.
* **Genesis Block**: The very first block ever mined and recorded on a blockchain network, hardcoded with index `0` and no previous block reference.
* **Immutability**: The quality of being unchangeable. In a blockchain, historical data is immutable because altering past blocks requires re-solving the computational proof for that block and all subsequent blocks.
* **Consensus Mechanism**: A fault-tolerant protocol used in blockchain networks to achieve agreement among distributed, untrusted nodes on the single, correct state of the ledger.
* **Byzantine Generals Problem**: A classic computer science problem modeling the difficulty of achieving consensus in a decentralized network where some participants may be corrupt or act maliciously.
